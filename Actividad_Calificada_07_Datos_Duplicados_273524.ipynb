## Actividad 1. Diagnóstico inicial (20 puntos)

### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
# duplicated() compara TODAS las columnas (subset=None): detecta duplicados exactos.
# .sum() cuenta los True (True = 1, False = 0).

# keep='first': marca todas las repeticiones excepto la primera aparición de cada grupo
n_first = hoteles.duplicated(keep='first').sum()

# keep='last': marca todas las repeticiones excepto la última aparición de cada grupo
n_last = hoteles.duplicated(keep='last').sum()

# keep=False: marca TODAS las filas que pertenecen a un grupo repetido, incluida la que se conservaría
n_todas = hoteles.duplicated(keep=False).sum()

print(f"keep='first': {n_first:,} filas sobrantes (copias de una fila que ya apareció antes)")
print(f"keep='last' : {n_last:,} filas sobrantes (copias de una fila que vuelve a aparecer después)")
print(f"keep=False  : {n_todas:,} filas involucradas en duplicados (todas las filas de cada grupo repetido)")
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# Porcentaje de filas sobrantes: la media de la máscara booleana es la proporción de True
porcentaje_sobrantes = hoteles.duplicated().mean() * 100

# Grupos repetidos: filas involucradas - filas sobrantes = una fila conservada por cada grupo
n_grupos = n_todas - n_first

print(f'Total de filas         : {len(hoteles):,}')
print(f'Filas sobrantes        : {n_first:,}')
print(f'Porcentaje de sobrantes: {porcentaje_sobrantes:.2f} %')
print(f'Grupos repetidos       : {n_grupos:,}')
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
# Copia para no modificar el DataFrame original
hoteles_dup = hoteles.copy()

# Columna booleana: True si la fila es una copia sobrante (keep='first')
hoteles_dup['es_duplicado'] = hoteles.duplicated(keep='first')

# Por hotel: size() da el total de reservas, sum() cuenta los True y mean() da su proporción
duplicados_hotel = pd.DataFrame({
    'reservas':   hoteles_dup.groupby('hotel').size(),
    'sobrantes':  hoteles_dup.groupby('hotel')['es_duplicado'].sum(),
    'porcentaje': hoteles_dup.groupby('hotel')['es_duplicado'].mean() * 100
})

print(duplicados_hotel)
```

## Actividad 2. Inspección y criterio de duplicado (25 puntos)

### 2.1 Ver los grupos repetidos (6 pts)

Filtra las filas involucradas en duplicados (`keep=False`), **ordénalas** con `sort_values()` para que las copias queden juntas y muestra las primeras 12 filas con estas columnas:

`hotel`, `lead_time`, `arrival_date_year`, `arrival_date_month`, `arrival_date_day_of_month`, `adults`, `country`, `market_segment`, `deposit_type`, `adr`, `reservation_status`.

Elige tú las columnas de ordenamiento y explica tu elección en un comentario.

```python
# Máscara con TODAS las filas que forman parte de un grupo repetido
mascara_dup = hoteles.duplicated(keep=False)

# Columnas que se mostrarán
columnas_vista = ['hotel', 'lead_time', 'arrival_date_year', 'arrival_date_month',
                  'arrival_date_day_of_month', 'adults', 'country', 'market_segment',
                  'deposit_type', 'adr', 'reservation_status']

# Criterio de ordenamiento: primero las columnas de la vista (hotel, anticipación, fecha de
# llegada, etc.), para que la tabla se lea fácil, y después el RESTO de las columnas.
# Como los duplicados son exactos (coinciden en las 32 columnas), ordenar por todas ellas
# garantiza que las copias de un mismo grupo queden en filas consecutivas; si ordenáramos
# solo por algunas, dos grupos que empatan en esas columnas podrían quedar intercalados.
columnas_orden = columnas_vista + list(hoteles.columns.drop(columnas_vista))

# Filtramos, ordenamos y mostramos las primeras 12 filas con las columnas pedidas
hoteles[mascara_dup].sort_values(by=columnas_orden)[columnas_vista].head(12)
```

### 2.2 Tamaño de los grupos (6 pts)

Usa `value_counts()` sobre el `DataFrame` completo para contar cuántas veces aparece cada combinación de filas. Considera que las columnas `agent` y `company` tienen faltantes (revisa el parámetro `dropna`).

Con ese resultado:

1. Muestra **cuántos grupos** hay de cada tamaño (2 copias, 3 copias, …). Basta con los 10 tamaños más frecuentes.
2. Indica el **tamaño del grupo más grande** y muestra su combinación de valores.

```python
# value_counts() sobre el DataFrame completo: cuenta cuántas veces aparece cada fila (combinación).
# dropna=False es necesario porque agent y company tienen NaN: con el valor por defecto
# (dropna=True) se descartan las combinaciones que contienen algún NaN.
tamanos = hoteles.value_counts(dropna=False)

# Comprobación del efecto de dropna: filas que quedan contadas con cada opción
print(f'Filas contadas con dropna=False: {tamanos.sum():,}')
print(f'Filas contadas con dropna=True : {hoteles.value_counts().sum():,}')

# 1) Solo grupos repetidos (aparecen más de una vez) y cuántos grupos hay de cada tamaño
print('\nNúmero de grupos según su tamaño (10 tamaños más frecuentes):')
print(tamanos[tamanos > 1].value_counts().head(10))

# 2) Tamaño del grupo más grande (value_counts ordena de mayor a menor)
print(f'\nTamaño del grupo más grande: {tamanos.max()} copias')
```

```python
# Combinación de valores del grupo más grande: value_counts() ordena de mayor a menor,
# así que es el primer elemento del índice de tamanos (una tupla con un valor por columna).
# zip() empareja cada nombre de columna con su valor para mostrar las 32 columnas completas
print(f'Combinación que se repite {tamanos.max()} veces:\n')
for columna, valor in zip(hoteles.columns, tamanos.index[0]):
    print(f'{columna:32}: {valor}')
```

Compara las filas involucradas en duplicados (`keep=False`) con el dataset completo en:

- Distribución de `market_segment` (proporciones).

## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
# drop_duplicates() sin subset: elimina duplicados exactos (todas las columnas);
# keep='first' conserva la primera aparición e ignore_index=True renumera el índice 0, 1, 2, ...
# Se asigna a una variable nueva para no modificar hoteles
hoteles_limpio = hoteles.drop_duplicates(keep='first', ignore_index=True)

print(f'Filas antes    : {len(hoteles):,}')
print(f'Filas después  : {len(hoteles_limpio):,}')
print(f'Filas eliminadas: {len(hoteles) - len(hoteles_limpio):,}')
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Medias por hotel de is_canceled (tasa de cancelación) y adr (tarifa diaria promedio)
resumen_original = hoteles.groupby('hotel')[['is_canceled', 'adr']].mean()
resumen_limpio = hoteles_limpio.groupby('hotel')[['is_canceled', 'adr']].mean()

# Número de reservas por hotel con size(); se alinea por el índice (hotel)
resumen_original['reservas'] = hoteles.groupby('hotel').size()
resumen_limpio['reservas'] = hoteles_limpio.groupby('hotel').size()

# reset_index() convierte el índice hotel en una columna normal
resumen_original = resumen_original.reset_index()
resumen_limpio = resumen_limpio.reset_index()

print('Original:')
print(resumen_original)
print('\nSin duplicados:')
print(resumen_limpio)
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
# Columna que identifica la versión de cada resumen
resumen_original['version'] = 'original'
resumen_limpio['version'] = 'sin duplicados'

# pd.concat() apila ambos resúmenes; se ordena por hotel (y por versión, para que
# 'original' quede siempre antes que 'sin duplicados') y se renumera el índice
comparacion = (
    pd.concat([resumen_original, resumen_limpio])
    .sort_values(by=['hotel', 'version'])
    .reset_index(drop=True)
)

comparacion
```

### 4.1 Verificación (5 pts)

Verifica sobre `hoteles_limpio` que:
