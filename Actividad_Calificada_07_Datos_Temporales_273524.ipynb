## Carga de datos

Ejecuta la siguiente celda. No requiere modificaciones.

```python
import pandas as pd

# Carga del dataset de taxis de Nueva York desde un repositorio público de GitHub.
# El archivo se lee con pandas; no se requiere ninguna otra biblioteca.
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/taxis.csv'
taxis = pd.read_csv(url)

print('Dimensiones:', taxis.shape)
taxis.head()
```

## Actividad 1. Conversión y validación (20 puntos)

### 1.1 Conversión (5 pts)

1. Muestra el tipo de dato de `pickup` y `dropoff`.
2. Convierte ambas columnas a tipo fecha con `pd.to_datetime()`, indicando el **formato explícito** que corresponde a los valores.
3. Verifica que la conversión no generó `NaT`: compara el número de faltantes antes y después.

```python
# 1) Tipo de dato original: al leer el CSV, pickup y dropoff llegan como texto
print('Tipos antes de convertir:')
print(taxis[['pickup', 'dropoff']].dtypes)

# Faltantes en el texto original, para compararlos después de la conversión
faltantes_antes = taxis[['pickup', 'dropoff']].isna().sum()

# 2) Los valores tienen la forma '2019-03-23 20:21:09': año-mes-día hora:minuto:segundo.
#    Indicamos ese formato explícito; con errors='coerce' cualquier valor que no lo cumpla
#    se convierte en NaT en lugar de detener el programa, y así podemos detectarlo.
formato = '%Y-%m-%d %H:%M:%S'
taxis['pickup'] = pd.to_datetime(taxis['pickup'], format=formato, errors='coerce')
taxis['dropoff'] = pd.to_datetime(taxis['dropoff'], format=formato, errors='coerce')

print('\nTipos después de convertir:')
print(taxis[['pickup', 'dropoff']].dtypes)

# 3) Verificación: si hay más NaT después que faltantes antes, la conversión perdió fechas
faltantes_despues = taxis[['pickup', 'dropoff']].isna().sum()
print('\nFaltantes antes de convertir:')
print(faltantes_antes)
print('\nNaT después de convertir:')
print(faltantes_despues)
```

### 1.2 Periodo de los datos (5 pts)

El dataset debería contener solo viajes de marzo de 2019.

1. Muestra la fecha mínima y máxima de `pickup` y de `dropoff`.
2. Cuenta y muestra los viajes cuyo `pickup` **no** está en marzo de 2019.
3. ¿Hay viajes con `pickup` en marzo y `dropoff` fuera de marzo? Cuéntalos.

```python
# 1) Rango de fechas de inicio y de término de los viajes
print('pickup  mínimo:', taxis['pickup'].min(), '| máximo:', taxis['pickup'].max())
print('dropoff mínimo:', taxis['dropoff'].min(), '| máximo:', taxis['dropoff'].max())

# 2) Máscara booleana: True si el viaje inicia en marzo de 2019 (año y mes con .dt)
pickup_en_marzo = (taxis['pickup'].dt.year == 2019) & (taxis['pickup'].dt.month == 3)

# Viajes cuyo pickup NO está en marzo de 2019 (negamos la máscara con ~)
fuera_periodo = taxis[~pickup_en_marzo]
print('\nViajes con pickup fuera de marzo de 2019:', len(fuera_periodo))
print(fuera_periodo[['pickup', 'dropoff', 'distance', 'fare', 'total']])

# 3) Viajes que inician en marzo pero terminan en otro mes o año
dropoff_en_marzo = (taxis['dropoff'].dt.year == 2019) & (taxis['dropoff'].dt.month == 3)
termina_fuera = taxis[pickup_en_marzo & ~dropoff_en_marzo]
print('\nViajes con pickup en marzo y dropoff fuera de marzo:', len(termina_fuera))
print(termina_fuera[['pickup', 'dropoff', 'distance', 'fare', 'total']])
```

2. Localiza `pickup` en la zona `'America/New_York'` con `tz_localize()` usando `nonexistent='NaT'` y cuenta cuántos valores quedan como `NaT`.

```python
# 1) Viajes que inician el 10 de marzo de 2019 (todo el día, ambos extremos incluidos)
dia_10 = taxis[taxis['pickup'].between('2019-03-10', '2019-03-10 23:59:59')]
print('Viajes con pickup el 10 de marzo:', len(dia_10))

# Número de viajes por hora de inicio; sort_index() las ordena de 0 a 23
print('\nViajes por hora el 10 de marzo:')
print(dia_10['pickup'].dt.hour.value_counts().sort_index())

# ¿Hay viajes en la hora que no existió (02:00 a 02:59)?
print('\nViajes con pickup a las 2:00:', (dia_10['pickup'].dt.hour == 2).sum())

# 2) Declaramos que pickup está en hora local de Nueva York. Las horas que no existieron
#    por el cambio de horario se marcan como NaT en lugar de producir un error.
#    Se guarda en una variable aparte para no modificar la columna original.
pickup_ny = taxis['pickup'].dt.tz_localize('America/New_York', nonexistent='NaT')
print('\nValores NaT al localizar en America/New_York:', pickup_ny.isna().sum())
```

## Actividad 2. Duraciones y anomalías (25 puntos)

### 2.1 Duración de cada viaje (6 pts)

1. Crea la columna `duracion` = `dropoff` − `pickup`.
2. Crea `duracion_min` con la duración en **minutos** (usa `.dt.total_seconds()`).
3. Muestra el resumen estadístico de `duracion_min` con `describe()`.

```python
# 1) Restar dos columnas datetime64 produce una duración (timedelta64)
taxis['duracion'] = taxis['dropoff'] - taxis['pickup']

# 2) Duración en minutos: total de segundos entre 60
taxis['duracion_min'] = taxis['duracion'].dt.total_seconds() / 60

print('Tipo de duracion:', taxis['duracion'].dtype)

# 3) Resumen estadístico de la duración en minutos
print(taxis['duracion_min'].describe())
```

### 2.2 Duraciones anómalas (6 pts)

Cuenta los viajes con:

- Duración **negativa** (termina antes de iniciar).
- Duración **cero**.
- Duración **menor a 1 minuto** (incluye los de duración cero).
- Duración **mayor a 3 horas**.

Muestra los viajes de duración cero con las columnas `pickup`, `dropoff`, `distance`, `fare` y `total`.

```python
# Cada condición compara la duración con un pd.Timedelta (una duración de referencia)
negativa = taxis['duracion'] < pd.Timedelta(0)                 # termina antes de iniciar
cero = taxis['duracion'] == pd.Timedelta(0)                    # inicia y termina en el mismo segundo
menor_1min = taxis['duracion'] < pd.Timedelta(minutes=1)       # incluye los de duración cero
mayor_3h = taxis['duracion'] > pd.Timedelta(hours=3)           # viajes excesivamente largos

print('Duración negativa  :', negativa.sum())
print('Duración cero      :', cero.sum())
print('Menor a 1 minuto   :', menor_1min.sum())
print('Mayor a 3 horas    :', mayor_3h.sum())

# Viajes de duración cero con sus datos de distancia y cobro
print('\nViajes de duración cero:')
print(taxis[cero][['pickup', 'dropoff', 'distance', 'fare', 'total']])

# Distancia y tarifa de los viajes de menos de 1 minuto (evidencia para la pregunta 2.4a)
menos_1min = taxis[menor_1min]
print('\nViajes de menos de 1 minuto con distancia 0 :', (menos_1min['distance'] == 0).sum())
print('Viajes de menos de 1 minuto con tarifa 2.50 :', (menos_1min['fare'] == 2.5).sum())
print('Viajes de menos de 1 minuto con tarifa >= 20:', (menos_1min['fare'] >= 20).sum())
print(menos_1min[['distance', 'fare']].describe())
```

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

```python
# 1) Velocidad en millas por hora: distancia entre la duración en horas.
#    Con duración cero la división no tiene sentido (división entre cero), así que
#    .where() conserva la velocidad solo donde la duración es mayor que cero y deja NaN en el resto.
velocidad = taxis['distance'] / (taxis['duracion_min'] / 60)
taxis['velocidad_mph'] = velocidad.where(taxis['duracion_min'] > 0)

print('Viajes sin velocidad (NaN):', taxis['velocidad_mph'].isna().sum())

# 2) Velocidades poco plausibles en tráfico urbano (más de 60 mph)
muy_rapidos = taxis['velocidad_mph'] > 60
print('Viajes con velocidad mayor a 60 mph:', muy_rapidos.sum())

# Cuántos de ellos NO se explican por durar menos de 1 minuto (evidencia para 2.4b)
print('De ellos, con duración de al menos 1 minuto:', (muy_rapidos & (taxis['duracion_min'] >= 1)).sum())

# 3) Los 5 viajes más rápidos, ordenados de mayor a menor velocidad
print('\nLos 5 viajes con mayor velocidad:')
print(taxis.sort_values(by='velocidad_mph', ascending=False).head(5)[
    ['pickup', 'dropoff', 'duracion_min', 'distance', 'velocidad_mph', 'fare']])
```

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# Componentes de pickup con el accesor .dt
taxis['hora'] = taxis['pickup'].dt.hour                  # hora de inicio (0 a 23)
taxis['dia_num'] = taxis['pickup'].dt.dayofweek          # lunes = 0, ..., domingo = 6
taxis['dia_nombre'] = taxis['pickup'].dt.day_name()      # nombre del día (en inglés)
taxis['fin_semana'] = taxis['dia_num'] >= 5              # sábado (5) y domingo (6)
taxis['fecha'] = taxis['pickup'].dt.date                 # solo la fecha, sin hora

# Un viaje cruza la medianoche si la fecha de término es distinta de la de inicio
taxis['cruza_medianoche'] = taxis['dropoff'].dt.date != taxis['fecha']

print(taxis[['pickup', 'hora', 'dia_num', 'dia_nombre', 'fin_semana', 'fecha', 'cruza_medianoche']].head())
print('\nViajes que cruzan la medianoche:', taxis['cruza_medianoche'].sum())
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
# Número de viajes y mediana de la duración para cada hora de inicio
viajes_hora = taxis.groupby('hora').size()
mediana_hora = taxis.groupby('hora')['duracion_min'].median()

# Reunimos ambos resultados en una tabla (el índice común es la hora)
patron_hora = pd.DataFrame({'viajes': viajes_hora, 'mediana_duracion_min': mediana_hora.round(1)})
print(patron_hora)

# Al ordenar de menor a mayor, el primer índice es el mínimo y el último el máximo
print('\nHora con más viajes  :', viajes_hora.sort_values().index[-1], 'h |', viajes_hora.max(), 'viajes')
print('Hora con menos viajes:', viajes_hora.sort_values().index[0], 'h |', viajes_hora.min(), 'viajes')

# Hora con la mayor duración mediana (necesaria para la pregunta 3.4a)
print('Hora con mayor duración mediana:', mediana_hora.sort_values().index[-1], 'h |',
      round(mediana_hora.max(), 1), 'min')
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
# Agrupamos por número y nombre del día: dia_num ordena de lunes (0) a domingo (6)
por_dia = taxis.groupby(['dia_num', 'dia_nombre'])

viajes_dia = por_dia.size()                          # número de viajes
mediana_duracion = por_dia['duracion_min'].median()  # mediana de la duración
mediana_tarifa = por_dia['fare'].median()            # mediana de la tarifa
fechas_distintas = por_dia['fecha'].nunique()        # cuántas fechas distintas tiene cada día de la semana

patron_dia = pd.DataFrame({
    'viajes': viajes_dia,
    'fechas': fechas_distintas,
    # Promedio por fecha: corrige que algunos días de la semana aparecen más veces en el mes
    'viajes_por_fecha': (viajes_dia / fechas_distintas).round(1),
    'mediana_duracion_min': mediana_duracion.round(1),
    'mediana_fare': mediana_tarifa
})
print(patron_dia)
```

```python
# Condiciones de exclusión del criterio definido en 2.4 (True = el viaje se excluye)
excl_periodo = (taxis['pickup'].dt.year != 2019) | (taxis['pickup'].dt.month != 3)   # inicia fuera de marzo de 2019
excl_min = taxis['duracion'] < pd.Timedelta(minutes=1)                              # dura menos de 1 minuto (incluye cero y negativas)
excl_max = taxis['duracion'] > pd.Timedelta(hours=3)                                # dura más de 3 horas
excl_vel = taxis['velocidad_mph'] > 60                                              # velocidad poco plausible (NaN no cuenta aquí)

# Viajes que incumple cada condición (un viaje puede incumplir más de una)
print('Excluidos por pickup fuera de marzo de 2019:', excl_periodo.sum())
print('Excluidos por duración menor a 1 minuto    :', excl_min.sum())
print('Excluidos por duración mayor a 3 horas     :', excl_max.sum())
print('Excluidos por velocidad mayor a 60 mph     :', excl_vel.sum())

# Dataset limpio: se excluye un viaje si cumple AL MENOS una condición (|) y se conservan los demás (~)
excluir = excl_periodo | excl_min | excl_max | excl_vel
taxis_limpio = taxis[~excluir].copy()

# El total no es la suma de las filas anteriores porque las condiciones se traslapan
print('\nViajes originales:', len(taxis))
print('Viajes excluidos en total:', len(taxis) - len(taxis_limpio))
print('Viajes en taxis_limpio  :', len(taxis_limpio))

# Verificación: en taxis_limpio ninguna fila debe incumplir alguna condición (todas en 0)
print('\nVerificación en taxis_limpio (filas que incumplen cada condición):')
print('Fuera de marzo de 2019 :', ((taxis_limpio['pickup'].dt.year != 2019) | (taxis_limpio['pickup'].dt.month != 3)).sum())
print('Menos de 1 minuto      :', (taxis_limpio['duracion'] < pd.Timedelta(minutes=1)).sum())
print('Más de 3 horas         :', (taxis_limpio['duracion'] > pd.Timedelta(hours=3)).sum())
print('Más de 60 mph          :', (taxis_limpio['velocidad_mph'] > 60).sum())
print('Velocidad faltante     :', taxis_limpio['velocidad_mph'].isna().sum())
print('Rango de pickup        :', taxis_limpio['pickup'].min(), '→', taxis_limpio['pickup'].max())
print('Rango de duracion_min  :', round(taxis_limpio['duracion_min'].min(), 2), '→', round(taxis_limpio['duracion_min'].max(), 2))
```

1. ¿Qué ventajas tuvo indicar el formato explícito en 1.1 frente a dejar que pandas lo infiera? ¿En qué situación la inferencia habría producido errores silenciosos?
2. Supón que `pickup` se hubiera registrado en UTC y `dropoff` en hora local de Nueva York. ¿Qué efecto tendría en las duraciones calculadas? ¿Cómo lo habrías detectado con las herramientas de esta actividad?
3. Si se entrenara un modelo para predecir la duración de un viaje, ¿qué problemas causarían las anomalías temporales que encontraste si no se limpiaran?

**Respuesta final:**

**1. Formato explícito vs. inferencia.** En este dataset las fechas vienen en un formato estándar (`2019-03-23 20:21:09`), así que la inferencia probablemente también habría funcionado. Indicar `'%Y-%m-%d %H:%M:%S'` tiene ventajas: documenta qué se espera de los datos, obliga a que **todos** los valores cumplan ese formato (con `errors='coerce'`, cualquier valor distinto se vuelve `NaT` y se detecta en la comparación antes/después, que aquí dio 0 y 0) y es más rápido en archivos grandes. La inferencia habría producido errores silenciosos si las fechas vinieran en un formato ambiguo día/mes, como `10/03/2019 01:23` (convención de México). pandas supone mes/día cuando puede, así que leería **3 de octubre** en lugar de 10 de marzo, mientras que un valor como `13/03/2019` sí se leería como 13 de marzo porque no existe el mes 13. Con `format='mixed'`, la columna quedaría con dos convenciones mezcladas sin ningún aviso: solo los días del 1 al 12 se confunden. Aquí afectaría justamente al 10 de marzo, el día del cambio de horario, y aparecerían viajes fuera del periodo que en realidad no lo están.

**2. `pickup` en UTC y `dropoff` en hora local.** En marzo de 2019, Nueva York estaba en UTC−5 hasta el 10 de marzo y en UTC−4 después. Si `pickup` estuviera en UTC, cada inicio aparecería 5 (o 4) horas **después** de la hora local real, y la duración calculada sería la real **menos 5 o 4 horas**. Como la duración máxima real es de 107.7 minutos (menos de 4 horas), **todos** los viajes tendrían duración negativa. Se habría detectado con varias herramientas de esta actividad:
   - En 2.2, el conteo de duraciones negativas habría sido de 6,433 en lugar de 0, y `describe()` habría mostrado una mediana negativa en lugar de 10.9 min.
   - En 1.3, habría viajes a las 2:00 del 10 de marzo y `tz_localize(..., nonexistent='NaT')` habría devuelto `NaT` en lugar de 0.
   - En 1.2, el `pickup` máximo pasaría del 31 de marzo a la madrugada del 1 de abril.
   - En 3.2, el patrón por hora estaría desplazado: el pico de las 18:00 aparecería a las 22:00 o 23:00, y el mínimo de las 5:00 hacia las 9:00 o 10:00, algo incoherente con el comportamiento de los usuarios.

**3. Efecto de no limpiar en un modelo que predice la duración.**
   - Los **72 viajes de menos de 1 minuto** enseñarían al modelo relaciones falsas: viajes de hasta 10.1 millas o tarifas de hasta 120 USD con duración casi cero. Esto contradice la relación esperada entre distancia y tiempo y empeora las predicciones para los viajes normales.
   - Las **velocidades imposibles** (hasta 4,834 mph) son valores extremos que tienen mucho peso en métricas basadas en la media, como el error cuadrático. Ya en los datos crudos la media de la duración (14.35 min) supera a la mediana (10.9 min) por la asimetría.
   - El **viaje del 28 de febrero** agregaría una fecha fuera del periodo y alteraría las variables de calendario (por ejemplo, el jueves quedó con 5 fechas en lugar de 4).
   - Si la **zona horaria** no se validara, variables como la hora del día o el fin de semana (que en 3.2, 3.3 y 4.2 muestran diferencias claras en la duración) quedarían desplazadas, y el modelo aprendería patrones que no corresponden a la realidad.
